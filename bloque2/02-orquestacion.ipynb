# Bloque 2 — Orquestación: Prompt Engineering, Frameworks y Agentes

> **Carga:** HT: 6h \| HPD: 4h
>

> [CE-2](../index.qmd#competencias-específicas),
> [CE-3](../index.qmd#competencias-específicas)

## Objetivos del bloque

Diseñar pipelines de prompting robustos, usar frameworks de orquestación
(LangChain, LlamaIndex), implementar servidores MCP y construir agentes
inteligentes con herramientas externas.

## Mapa conceptual del bloque



Este bloque te lleva de la interacción básica con un LLM hasta la
construcción de sistemas autónomos capaces de razonar, usar herramientas
y ejecutar tareas multi-paso. Empezamos refinando el arte del prompt,
luego aprendemos a orquestar con frameworks, estandarizamos la conexión
a datos con MCP y finalmente construimos agentes inteligentes.

## Ruta de aprendizaje recomendada



1.  Comienza con **Prompt Engineering** para dominar cómo comunicarte
    con el LLM
2.  Aprende **LangChain** o **LlamaIndex** (o ambos, son
    complementarios) para orquestar
3.  Entiende **MCP** para estandarizar la conexión a datos
4.  Construye **agentes** uniendo todos los conceptos
5.  Ponlo en práctica con los HPDs: un **agente con herramientas**
    (HPD 2) y un **servidor MCP** (HPD 3)
