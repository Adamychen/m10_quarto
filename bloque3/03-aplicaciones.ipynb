> **Carga:** HT: 6h \| HPD: 4h

## Objetivos del bloque

Implementar sistemas RAG de calidad producción con evaluación objetiva,
generar datos sintéticos validados estadísticamente, y aplicar técnicas
de ajuste fino eficientes (LoRA/QLoRA) para adaptar modelos a dominios
específicos.

## Mapa conceptual del bloque



Los tres pilares de este bloque son **complementarios y se refuerzan
mutuamente**:

- **RAG Avanzado** te enseña a construir sistemas de pregunta-respuesta
  robustos, con evaluación objetiva (RAGAS) y técnicas avanzadas
  (re-ranking, query expansion, multi-fuente).
- **Datos Sintéticos** te permite generar datos de entrenamiento,
  testing o anonimización cuando los datos reales son insuficientes o
  sensibles.
- **Fine-tuning** te permite adaptar modelos open-source a tu dominio
  específico cuando RAG y prompting no bastan.

## Ruta de aprendizaje recomendada



1.  Empieza con **RAG Avanzado**: es el tema central del bloque y la
    base del HPD 4. Aprenderás a mejorar la calidad de tus sistemas RAG
    con re-ranking, query expansion y evaluación objetiva con RAGAS.
2.  Continúa con **Datos Sintéticos**: aprenderás a generar datos
    verosímiles con LLMs (enfoque intuitivo) y con modelos
    especializados (CTGAN, TVAE).
3.  Cierra con **Fine-tuning**: entenderás cuándo y cómo adaptar modelos
    open-source a tu dominio con LoRA/QLoRA.
4.  Ponlo todo en práctica con el **HPD 4**: construirás un sistema RAG
    completo sobre datos mixtos (PDFs + SQL) con evaluación RAGAS.
