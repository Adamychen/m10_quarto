# Bloque 3 — Aplicaciones: RAG Avanzado, Datos Sintéticos y Fine-tuning

> **Competencias asociadas:**
> [CE-4](../index.qmd#sec-competencias-especificas),
> [CE-5](../index.qmd#sec-competencias-especificas)

## Objetivos del bloque

Implementar sistemas RAG de calidad de producción con evaluación
objetiva, generar datos sintéticos validados estadísticamente y aplicar
técnicas de ajuste fino eficientes (LoRA/QLoRA) para adaptar modelos a
dominios específicos.

## Mapa conceptual del bloque

<figure id="fig-b3-mapa-conceptual">

<figcaption>Figure 1</figcaption>
</figure>

Los tres pilares de este bloque son complementarios y se refuerzan
mutuamente:

- **RAG Avanzado** te enseña a construir sistemas de pregunta-respuesta
  robustos, con evaluación objetiva (RAGAS) y técnicas avanzadas
  (re-ranking, query expansion, multi-fuente).
- **Datos Sintéticos** te permite generar datos de entrenamiento,
  pruebas o anonimización cuando los datos reales son insuficientes o
  sensibles.
- **Fine-tuning** te permite adaptar modelos de código abierto a tu
  dominio específico cuando RAG y prompting no bastan.

## Ruta de aprendizaje recomendada

<figure id="fig-b3-ruta-aprendizaje">

<figcaption>Figure 2</figcaption>
</figure>

1.  Empieza con **RAG Avanzado**: es el tema central del bloque y la
    base del HPD 5. Aprenderás a mejorar la calidad de tus sistemas RAG
    con re-ranking, query expansion y evaluación objetiva con RAGAS.
2.  Continúa con **Datos Sintéticos**: aprenderás a generar datos
    verosímiles con LLM (enfoque intuitivo) y con modelos especializados
    (CTGAN, TVAE).
3.  Cierra con **Fine-tuning**: entenderás cuándo y cómo adaptar modelos
    open-source a tu dominio con LoRA/QLoRA.
4.  Ponlo todo en práctica con el **HPD 5**: construirás un sistema RAG
    completo sobre datos mixtos (PDFs + SQL) con evaluación RAGAS.

## Resumen del bloque

<figure id="fig-b3-resumen">

<figcaption>Figure 3</figcaption>
</figure>

Este bloque te da las herramientas para llevar tus sistemas IA a calidad
de producción:

- **Bloque 4:** evaluarás y desplegarás estos sistemas con gobernanza,
  detectando alucinaciones y sesgos, y documentando con Model Cards.
- **Proyecto integrador:** combinarás RAG + datos sintéticos +
  fine-tuning en un sistema completo de tu dominio.

Ahora pasa al primer tema: **[RAG Avanzado](03b-rag-avanzado.qmd)**.
