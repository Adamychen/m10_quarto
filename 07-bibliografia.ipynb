Esta página agrupa los recursos externos que complementan los materiales
del módulo: documentación de herramientas, lecturas de contexto y un
índice curado de Model/System Cards. La bibliografía completa de los
papers y fuentes citadas en el sitio se genera automáticamente al final
de esta página desde el archivo `references.bib`.

## Historia de las representaciones y embeddings

Estas lecturas acompañan el [panorama histórico del Bloque
1](bloque1/00-panorama.qmd) y permiten seguir el paso desde las
representaciones dispersas hasta la recuperación semántica:

- Bengio et al. (2003) — modelos neuronales de lenguaje y
  representaciones distribuidas.
- Mikolov et al. (2013) — embeddings de palabras mediante CBOW y
  Skip-gram.
- Pennington et al. (2014) — vectores globales a partir de estadísticas
  de coocurrencia.
- Peters et al. (2018) — representaciones contextuales con ELMo.
- Reimers & Gurevych (2019) — embeddings de frases con arquitectura
  siamesa.
- Wang et al. (2022) — aprendizaje contrastivo para embeddings de texto.
- Karpukhin et al. (2020) — recuperación densa de pasajes.
- Lewis et al. (2020) — generación aumentada mediante recuperación.
