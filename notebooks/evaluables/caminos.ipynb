# Elegir tu camino

> Al inicio del módulo, cada alumno elige uno de los tres caminos de
> aplicación. Esta elección determina el contexto, los datasets y la
> narrativa de los entregables HPD 1-5.
>
> **Las competencias y los criterios de evaluación son los mismos** en
> los tres caminos. Solo cambia el dominio de aplicación.
>
> **El proyecto integrador final es libre**: los caminos A/B/C aplican a
> los HPD 1-5. El proyecto final se define con tu propio problema y tu
> propio patrón agéntico (ver [Proyecto integrador
> final](proyecto-final.qmd)).

## Los tres caminos

| Camino | Contexto | Datos principales | Perfil de salida |
|------------------|------------------|------------------|------------------|
| **A — Finanzas** | Análisis financiero y corporativo | Informes anuales, balances, cotizaciones, noticias de mercado | Data Scientist financiero |
| **B — Salud** | Literatura biomédica y datos clínicos | Artículos PubMed, ensayos clínicos, historiales sintéticos | Data Scientist biomédico |
| **C — E-commerce** | Comercio electrónico y productos | Reviews de productos, catálogos, datos de ventas, customer support | Data Scientist de producto |

Table 1: Resumen de los tres caminos del módulo

### Camino A — Finanzas

Trabajarás con datos del mundo financiero: informes anuales de empresas,
balances, cotizaciones históricas y noticias de mercado. Construirás
sistemas que ayuden a analistas a responder preguntas como:

- *“¿Qué factores menciona el informe anual de 2024 como riesgos
  principales?”*
- *“¿Qué productos tienen margen inferior al 20 % según los datos del
  Q2?”*
- *“¿Cómo ha evolucionado la deuda neta en los últimos 3 trimestres?”*

**Datasets principales:** informes financieros generados sintéticamente,
tablas de balances y cuentas de resultados, series temporales de
cotizaciones.

### Camino B — Salud

Trabajarás con datos del ámbito biomédico: abstracts de PubMed, ensayos
clínicos, protocolos de tratamiento e historiales de pacientes
sintéticos. Construirás sistemas que ayuden a profesionales clínicos a
responder preguntas como:

- *“¿Qué efectos secundarios se mencionan en los ensayos de este
  fármaco?”*
- *“¿Qué pacientes tienen factores de riesgo compatibles con este
  perfil?”*
- *“¿Cuál es la dosis recomendada según los protocolos más recientes?”*

**Datasets principales:** abstracts de artículos biomédicos, tablas de
ensayos clínicos, historiales de pacientes sintéticos.

### Camino C — E-commerce

Trabajarás con datos de comercio electrónico: catálogos de productos,
reviews de clientes, datos de ventas e interacciones de soporte.
Construirás sistemas que ayuden a equipos de producto a responder
preguntas como:

- *“¿Qué aspectos del producto destacan las reviews positivas?”*
- *“¿Qué productos tienen mayor tasa de retorno por categoría?”*
- *“¿Cuál es el sentimiento general hacia esta línea de productos?”*

**Datasets principales:** catálogos de productos, reviews de clientes,
tablas de ventas e inventario, tickets de soporte sintéticos.

## Cómo elegir

Elige el camino que más se alinee con tus intereses profesionales o
académicos. Los tres caminos cubren las mismas competencias técnicas; la
diferencia está en el dominio de aplicación. Todos los entregables del
curso seguirán el camino elegido.

``` python
CAMINO = "A"  # Cambia a "B" o "C" según tu elección
print(f"Has elegido el Camino {CAMINO}.")
```

Los scripts de corrección leerán la variable `CAMINO` para interpretar
las soluciones de cada entregable.

## Ruta completa por camino

<figure id="fig-ruta-caminos">

<figcaption>Figure 1</figcaption>
</figure>

> **¿Y el proyecto final?**
>
> Los caminos A/B/C te dan datos y contexto para los HPD 1-5. El
> [proyecto integrador final](proyecto-final.qmd) es otra cosa: eliges
> tu propio problema y aplicas un patrón de diseño de agentes de la
> taxonomía de [02f — Patrones de Diseño de
> Agentes](../bloque2/02f-patrones-agentes.qmd) — o propones uno propio.
> Puedes reutilizar los datos de tu camino como punto de partida, pero
> el proyecto no está limitado a ellos.
