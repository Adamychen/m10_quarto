# Evaluación

## Estructura de la evaluación

| Instrumento | Peso | Nota mínima | Vinculación |
|------------------|------------------|------------------|------------------|
| Entregas prácticas parciales (HPD 1, 2, 4 y 5) | 30 % | 5,0 / 10 | CE-1, CE-2, CE-3, CE-4, CE-5 |
| Participación en foros (mínimo 3 de 4) | 10 % | Participado / No participado | CE-6 (pensamiento crítico) |
| Proyecto integrador final | 60 % | 5,0 / 10 | CE-1 a CE-6 |

Table 1: Instrumentos de evaluación y su peso en la nota final

## Contexto de los entregables

Los entregables HPD 1-5 comparten un único contexto de aplicación:
**análisis financiero y corporativo** (informes, balances, noticias de
mercado, datos de crédito). Las competencias y los criterios de
evaluación son los de siempre; el contexto único evita que cada alumno
trabaje con datos distintos y facilita la corrección. El proyecto
integrador final es libre: problema y patrón agéntico propios (ver
sección siguiente).

## Proyecto integrador final

El proyecto integrador final es 100 % libre: cada alumno elige su propio
problema y aplica un patrón de diseño de agentes de la taxonomía de [02f
— Patrones de Diseño de Agentes](bloque2/02f-patrones-agentes.qmd) (o
propone uno propio, con bonus). Los detalles completos —propuesta, ruta
de datos, rúbrica, criterios de éxito y formato de entrega— se
encuentran en la página dedicada:

<a href="evaluables/proyecto-final.qmd"
class="btn btn-primary"><strong>Proyecto integrador final →</strong></a>

Allí encontrarás:

- La fase de propuesta (`propuesta.md`): problema, criterios de éxito
  medibles, patrón candidato y ruta de datos
- Las tres rutas de datos: dataset real abierto, datos sintéticos
  generados por ti, o corpus propio
- La rúbrica con la dimensión “Patrón” como eje (justificación,
  aplicación y comparativa contra baseline)
- El bonus por idea novedosa: proponer un patrón propio o variante
- Los criterios de éxito, el formato de entrega y consejos para abordar
  el proyecto
